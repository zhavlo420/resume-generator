```python
import requests
import gradio as gr
from fpdf import FPDF
```

```python
OLLAMA_URL= ""
def gen_resume(name , job_role ,experience,skills, education , summary):

  prompt= f"generate a pro resume based on the details"\
          f"name: {name}\n"\
          f"job role: {job_role}\n"\
          f"experience: {experience}\n"\
          f"skills: {skills}\n"\
          f"education: {education}\n"\
          f"summary: {summary}"

  payload = {
      "model" : "deepseek-r1",
      "prompt" : prompt,
      "stream": False
  }

  response= requests.post(OLLAMA_URL,json=payload)

  if response.status_code ==200:
    return response.json().get("response","no res")
  else:
      return f"Error {response.text}"

  if __name__ == "__main__":
    test= generate_resume("john doe","swe","3","frontend,backend","bcom","exp in fullstack")
    print(test)
```

```python
interface = gr.Interface(
    fn=gen_resume,
    inputs=[
        gr.Textbox(label="Name"),
        gr.Textbox(label="Job Role"),
        gr.Slider(minimum=0, maximum=50, step=1, label="Years of Experience"),
        gr.Textbox(label="Skills (Comma separated)"),
        gr.Textbox(label="Education"),
        gr.Textbox(label="Summary")
    ],
    outputs=gr.Textbox(label="Resume")
)
interface.launch()
```
